### Using the Client ###
The ```mlflow.client``` module provides a CRUD interface to ```MLflow Experiments```, ```Runs```, ```Model Versions```, and ```Registered Models```. It's a client of:
- an ```MLflow Tracking Server``` that creates and manages ```experiments``` and ```runs```;
- an ```MLflow Registry Server``` that creates and manages registered ```models``` and ```model versions```. 

It’s a thin wrapper around ```TrackingServiceClient``` and ```RegistryClient``` so there is a unified API but also allowing the implementation of the tracking and registry clients independent from each other.

````python
# Instantiating the 'MlflowClient()'
client = mlflow.client.MlflowClient(tracking_uri= "http://127.0.0.1:8080")
client
````

**Creating Experiment**

````python
# Creating an experiment
experiment_id = client.create_experiment(name= "tutorial_client", # 'name': experiment name (must be unique)
                                         artifact_location= "./mlruns", # 'artifact_location': location of 'mlruns' folder (or where to store artifacts)
                                         tags= {"source": "client"}) # 'tags': dictionary of key-value pairs
experiment_id
````

````python
# Setting tags
client.set_experiment_tag(experiment_id= experiment_id, # 'experiment_id': ID of experiment
                          key= "api_tag", # 'key': tag name
                          value= True) # 'value': tag value (converts to str)
````

**Experiment**

````python
# Getting the experiment by ID
experiment = client.get_experiment(experiment_id= experiment_id) # 'experiment_id': ID of experiment
# Experiment information
print(f"Artifact Location: {experiment.artifact_location}")
print(f"Created At: {experiment.creation_time}")
print(f"Experiment ID: {experiment.experiment_id}")
print(f"Updated At: {experiment.last_update_time}")
print(f"Lifecycle Stage: {experiment.lifecycle_stage}")
print(f"Name: {experiment.name}")
print(f"Tag: {experiment.tags}")
````

````python
# Getting the experiment by name
client.get_experiment_by_name(name= "tutorial_client") # 'name': experiment name (case sensitive)
                                                       # Same output as 'get_experiment()'
````

**Model Version**

````python
# Getting registered model versions (to be deprecated)
model_versions = client.get_latest_versions(name= "tutorial_churn") # 'name': name of registered model
# Ouputting
for each in model_versions:
    print(f"Aliases: {each.aliases}")
    print(f"Created At: {each.creation_timestamp}")
    print(f"Current Stage: {each.current_stage}")
    print(f"Description: {each.description}")
    print(f"Updated At: {each.last_updated_timestamp}")
    print(f"Name: {each.name}")
    print(f"Run ID: {each.run_id}")
    print(f"Run Link: {each.run_link}")
    print(f"Source: {each.source}")
    print(f"Status: {each.status}")
    print(f"Status Message: {each.status_message}")
    print(f"Tags: {each.tags}")
    print(f"User ID: {each.user_id}")
    print(f"Version: {each.version}")
    print("-------------------------------------------------------------\n")
````

````python
# Getting registered model versions
client.get_model_version(name= "tutorial_churn", # 'name': name of registered model
                         version= 1) # 'version': model version (integer)
````

````python
# Getting registered model versions (by alias)
client.get_model_version_by_alias(name= "tutorial_churn", # 'name': name of registered model
                                  alias= "best") # 'alias': model alias
````

````python
# Getting model versions download URI
client.get_model_version_download_uri(name= "tutorial_churn", # 'name': name of registered model
                                      version= 1) # 'version': model version (integer)
````

**Registered Model**

````python
# Getting the registered model
client.get_registered_model(name= "tutorial_churn") # 'name': name of registered model
````

**Run**

````python
# Getting the run (and its metadata)
run = client.get_run(run_id= "72cb86a8bea44ff3b37e0a369c9e157d") # 'run_id': ID of run
# Ouputting
print(f"Metrics: {run.data.metrics}")
print(f"Parameters: {run.data.params}")
print(f"Tags: {run.data.tags}")
print(f"Run Info: {run.info}")
print(f"Inputs: {run.inputs}")
````

**Fetching Metric**

````python
# Getting list of metric objects (with all values logged)
client.get_metric_history(run_id= "72cb86a8bea44ff3b37e0a369c9e157d", # 'run_id': ID of run
                          key= "acc_train") # 'key': metric name in run
````
